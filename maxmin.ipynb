## Home work

WAP to find Maximum, Second Maximum, Minimum and Second Minimum numbers from a list using set() and for loop. Write separate programs for each.

1. Maximum

```python
l = [34,7,90,57,34,89,23,45,90,90,89,89]
s1 = set(l)             

max_num = 0              # To store largest number

for i in s1:             
    if i > max_num:      # If current number GT max_num, consider current number as maximum number 
        max_num = i      

print("Max number =", max_num)  
```

2. Second Maximum

```python
l = [34,7,90,57,34,89,23,45,90,90,89,89]
s1 = set(l)                  

max_num = 0                  # To store largest number
second_max = 0               # To store Second largest number

for i in s1:                 

    if i > max_num:          # If current number GT max_num, consider privious largest number as second maximum number
        second_max = max_num 
        max_num = i          

    elif i > second_max and i != max_num:
        # If current number GT and not equal to second max_num,

        second_max = i       # So current number beacame second maximum.

print("Second Max number =", second_max)
```

3. Minimum

```python
l = [34,7,90,57,34,89,23,45,90,90,89,89]
s1 = set(l)                  

min_num = None               # No minimum number present for now.

for i in s1:                 
    if min_num is None or i < min_num:
        # If min_num is empty or current number less than min_num

        min_num = i          # Current number became minimum number

print("Min number =", min_num)
```

4. Second Minimum

```python
l = [34,7,90,57,34,89,23,45,90,90,89,89]
s1 = set(l)                     

min_num = None                  # Store smallest number
second_min = None               # Store second smallest number

for i in s1:                    

    if min_num is None or i < min_num:
        # If minimum is empty now or current number is less than to minimum number

        second_min = min_num    # Previous minimum became second minimum
        min_num = i             # Current number became new minimum

    elif (second_min is None or i < second_min) and i != min_num:
        # If current number is less than to second_min and current number is not equal to minimum

        second_min = i          # Current number became second minimum

print("Second Min number =", second_min)
```
